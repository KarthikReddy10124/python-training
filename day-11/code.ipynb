```python
# 35. Merge two sorted lists
lst1 = [1,3,5]
lst2 = [2,4,6]
merged = []
i = j = 0
while i < len(lst1) and j < len(lst2):
    if lst1[i] < lst2[j]:
        merged.append(lst1[i])
        i += 1
    else:
        merged.append(lst2[j])
        j += 1
merged += lst1[i:] + lst2[j:]
print("Merged list:", merged)
```

```python
#17.Find most frequent 2 words.
s = input("Enter a sentence: ").lower()
words = s.split()
word_frequncy={}
for word in words:
    if word in word_frequncy:
        word_frequncy[word] += 1
    else:
        word_frequncy[word] = 1
for word, freq in word_frequncy.items():
    print(f'"{word}": {freq}')
sorted_words = sorted(word_frequncy.items(), key=lambda item: item[1], reverse=True)
most_frequent_2 = sorted_words[:1]
```

```python
# 41. Flatten nested list
nested = [1, [2, [3,4]], 5]
flat = []
def flatten(lst):
    for i in lst:
        if isinstance(i, list):
            flatten(i)
        else:
            flat.append(i)
flatten(nested)
print("Flattened list:", flat)
```
