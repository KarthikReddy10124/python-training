```python
n=int(input("Enter the number of terms:"))
a,b=0,1
count=0
if n<=0:
    print("enter a valid integer")
elif n==1:
    print("Fibonacci sequence upto",n,":")
    print(a)
else:
    print("Fibonaci sequence: ")
    while count<n:
        print(a, end=' ')
        c=a+b
        a=b
        b=c
        count+=1
```

```python
#18.sorting a random integer+string list in acceding order and descending order example input 
mixed_list = [9, 'c', 2, 'a', 5, 'b']
# Separate integers and strings
integers = [item for item in mixed_list if isinstance(item, int)]
strings = [item for item in mixed_list if isinstance(item, str)]
# Sort integers and strings separately
sorted_integers_asc = sorted(integers)
sorted_strings_asc = sorted(strings)
sorted_integers_desc = sorted(integers, reverse=True)
sorted_strings_desc = sorted(strings, reverse=True)
print("Sorted integers (ascending):", sorted_integers_asc)
print("Sorted strings (ascending):", sorted_strings_asc)
print("Sorted integers (descending):", sorted_integers_desc)
print("Sorted strings (descending):", sorted_strings_desc)
```
