```python
def total(*args):
    return sum(args)
def profile(**kwargs):
    return kwargs
print(total(2,4,6,9))
print(profile(name="karthik", role="SDE"))
```

```python
def number(n):
    for i in range(n):
        yield i
for x in number(3):
    print(x)
```
