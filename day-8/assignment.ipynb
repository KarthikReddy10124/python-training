```python
#1.palindrome -integer way

a = int(input("Enter a number: "))
num = a
rev = 0
while a!=0:
    rem = a%10
    rev = rev*10 + rem
    a = a//10
if rev == num:
    print(f'{num} is a palindrome')
else:
    print(f'{num} is not a palindrome')
```

```python
#2.palindrome -string way
s = input("Enter a string: ").lower()
if s == s[::-1]:
    print(f'"{s}" is a palindrome')
else:
    print(f'"{s}" is not a palindrome')
```

```python
#3.Fibbonacci series
n = int(input("Enter the number of terms: "))
a, b = 0, 1
count = 0
if n <= 0:
    print("Please enter a positive integer")
elif n == 1:
    print("Fibonacci sequence upto",n,":")
    print(a)
else:
    print("Fibonacci sequence:")
    while count < n:
        print(a, end=' ')
        c = a + b
        a = b
        b = c
        count += 1
```

```python
#4.Armstrong number
num = int(input("Enter a number: "))
order = len(str(num))
sum = 0
temp = num
while temp > 0:
    digit = temp % 10
    sum += digit ** order
    temp //= 10
if num == sum:
    print(f'{num} is an Armstrong number')
else:
    print(f'{num} is not an Armstrong number')
```

```python
#5.Reverse a string in 3 ways
s = input("Enter a string: ")
# Method 1: Using slicing
rev1 = s[::-1]
print("Reversed string (slicing):", rev1)
# Method 2: Using reversed() function
rev2 = ''.join(reversed(s))
print("Reversed string (reversed function):", rev2)
# Method 3: Using a loop
rev3 = ''
for char in s:
    rev3 = char + rev3
print("Reversed string (loop):", rev3)
```
